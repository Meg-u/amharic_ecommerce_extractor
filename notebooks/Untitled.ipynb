```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output
import re

# CONFIG 
CSV_PATH = '../data/fashiontera/messages_cleaned.csv'
SAVE_PATH = '../data/conll_dataset.txt'
NUM_TO_LABEL = 30
```

```python
df = pd.read_csv(CSV_PATH, encoding='utf-8', dtype=str)

# Show a few columns side by side if available
print(df.head(5).to_string())
```

```python
messages = df['text'].dropna().head(NUM_TO_LABEL).tolist()
```

```python
label_options = ['O', 'B-Product', 'I-Product', 'B-PRICE', 'I-PRICE', 'B-LOC', 'I-LOC', 'B-CONTACT_INFO']
labeled_data = []
```

```python
def custom_tokenize(text):
    text = text.replace('_', ' ')  # split compound tokens
    tokens = []

    # Keep Amharic (U+1200–137F), Latin (A-Za-z), digits, punctuation
    cleaned_text = re.sub(r'[^\u1200-\u137F\u0041-\u005A\u0061-\u007A0-9\s/@#:]', '', text)

    # Split on spaces and slashes
    rough_tokens = re.split(r'\s+|/', cleaned_text)

    # Remove empty tokens
    tokens = [tok for tok in rough_tokens if tok.strip()]

    return tokens
```

```python
def label_message(index):
    #clear_output(wait=True)

    if index >= len(messages):
        with open(SAVE_PATH, 'a', encoding='utf-8') as f:
            for entry in labeled_data:
                for token, label in entry:
                    f.write(f"{token} {label}\n")
                f.write("\n")
        print(f"All done! Labeled {len(messages)} messages and saved to {SAVE_PATH}")
        return

    msg = messages[index]
    tokens = custom_tokenize(msg)

    print(f"\n📨 Message {index + 1} / {len(messages)}")
    print("Original Message:")
    print(msg)
    print("\n Start labeling:")

    dropdowns = []
    for token in tokens:
        label_dropdown = widgets.Dropdown(
            options=label_options,
            value='O',
            description=token,
            layout=widgets.Layout(width='auto'),
            style={'description_width': 'initial'}
        )
        dropdowns.append(label_dropdown)
        display(label_dropdown)

    # Make sure this function is defined right here
    def save_labels(_):
        entry = [(dropdown.description, dropdown.value) for dropdown in dropdowns]
        labeled_data.append(entry)
        label_message(index + 1)  # Recursive call to next message

    # Create and bind the button AFTER defining save_labels
    save_button = widgets.Button(description="Save & Next", button_style='success')
    save_button.on_click(save_labels)
    display(save_button)
label_message(2)
```

```python
with open(SAVE_PATH, 'a', encoding='utf-8') as f:
    for entry in labeled_data:
        for token, label in entry:
            f.write(f"{token} {label}\n")
        f.write("\n")  # blank line between messages
```
